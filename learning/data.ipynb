# Embeddings

```python
uniq_words = df["text"].unique()
with open("models/embedding/uniq_words.txt", 'w', encoding='UTF-8') as f:
    for word in uniq_words:
        f.write(f"{word}\n")
```

```python
import fasttext.util
import numpy as np
import os
download_path = "./models/embedding/fasttext-ru"
# original_dir = os.getcwd()

# try:
#     os.chdir(download_path)
#     fasttext.util.download_model('ru', if_exists='ignore')
# finally:
#     os.chdir(original_dir)

embedding_model = fasttext.load_model(f"{download_path}/cc.ru.300.bin")
```

```python
def similarity(v1, v2):
    return np.dot(v1, v2) / (np.linalg.norm(v1) * np.linalg.norm(v2))
```

```python
embedding_data = []

for word in uniq_words:
    vec = embedding_model.get_word_vector(word)
    embedding_data.append(vec / np.linalg.norm(vec))

embedding_df = pd.DataFrame(embedding_data, index=uniq_words)
embedding_df
```

```python
embedding_df.to_csv("models/embedding/word_vectors.csv", encoding="UTF-8")
```

```python
embedding_df = pd.read_csv("models/embedding/word_vectors.csv", encoding="UTF-8", index_col=0)
uniq_words = df["text"].unique()
embedding_matrix = embedding_df.values.astype('float32')
embedding_dict = {}
embedding_dict_reverse = {}
for i, word in enumerate(embedding_df.index):
    embedding_dict[word] = i
    embedding_dict_reverse[i] = word
```

```python
norms = []
for word_emb in embedding_matrix:
    norms.append(np.linalg.norm(word_emb))

print(min(norms))
print(max(norms))
print(sum(norms) / len(norms))
```

```python
similarities = [(0, '', '') for i in range(1000000)]
ind = 0

for i in range(len(uniq_words)):
    for j in range(i+1, len(uniq_words)):
        vector1 = embedding_df.loc[uniq_words[i]]
        vector2 = embedding_df.loc[uniq_words[j]]
        sim = similarity(vector1, vector2)
        similarities[ind] = (sim, uniq_words[i], uniq_words[j])
        ind += 1

similarities = sorted(similarities, reverse=True,
                      key = lambda x : x[0])           
```
